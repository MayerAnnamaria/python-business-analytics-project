````python

# =============================================================================
#  PUBLISH BUSINESS ANALYTICS REPORTS TO GITHUB
# =============================================================================
# Purpose:
#   1. Configure Git identity and GitHub authentication.
#   2. Locate generated business analytics reports.
#   3. Stage and commit CSV reports.
#   4. Push the reports to the GitHub repository.
#   5. Confirm successful publication.
# =============================================================================

from pathlib import Path
from getpass import getpass
import subprocess
import os

# =============================================================================
# 01 | USER CONFIGURATION — EDIT ONLY THIS SECTION
# =============================================================================

#USERNAME = "your_github_username"
#EMAIL = "your_github_email@example.com"
#REPO_NAME = "python-business-analytics-project"
#BRANCH = "main"

# Enter your GitHub Personal Access Token securely when prompted.
TOKEN = getpass("Enter your GitHub Personal Access Token: ")

# =============================================================================
# 02 | DEFINE PROJECT DIRECTORIES
# =============================================================================

PROJECT_DIR = Path("/content") / REPO_NAME
OUTPUT_DIR = PROJECT_DIR / "outputs"

# Verify the local Git repository exists.
if not (PROJECT_DIR / ".git").exists():
    raise FileNotFoundError(f"Git repository not found: {PROJECT_DIR}")

# =============================================================================
# 03 | CONFIGURE GIT IDENTITY
# =============================================================================

# Identify the author of commits made from this Colab environment.
subprocess.run(
    ["git", "config", "user.name", USERNAME],
    cwd=PROJECT_DIR,
    check=True
)

subprocess.run(
    ["git", "config", "user.email", EMAIL],
    cwd=PROJECT_DIR,
    check=True
)

# =============================================================================
# 04 | LOCATE GENERATED BUSINESS REPORTS
# =============================================================================

csv_files = sorted(OUTPUT_DIR.glob("*.csv"))

if not csv_files:
    raise FileNotFoundError(
        f"No CSV reports found in: {OUTPUT_DIR}"
    )

print("\nGENERATED BUSINESS REPORTS")
print("-" * 50)

for file in csv_files:
    print(f"- {file.name}")

# =============================================================================
# 05 | STAGE REPORTS FOR VERSION CONTROL
# =============================================================================

# Stage only CSV reports, including files excluded by .gitignore.
subprocess.run(
    ["git", "add", "-f", "--"] +
    [str(file.relative_to(PROJECT_DIR)) for file in csv_files],
    cwd=PROJECT_DIR,
    check=True
)

# Display a summary of the files prepared for commit.
print("\nSTAGED CHANGES")

subprocess.run(
    ["git", "diff", "--cached", "--stat"],
    cwd=PROJECT_DIR,
    check=True
)

# =============================================================================
# 06 | COMMIT REPORTS
# =============================================================================

# Check whether there are staged changes to commit.
changes = subprocess.run(
    ["git", "diff", "--cached", "--quiet"],
    cwd=PROJECT_DIR
)

if changes.returncode == 1:

    subprocess.run(
        [
            "git",
            "commit",
            "-m",
            "Update business analytics reports"
        ],
        cwd=PROJECT_DIR,
        check=True
    )

    print("\nReports committed successfully.")

elif changes.returncode == 0:
    print("\nReports are already committed. No new changes.")

else:
    raise RuntimeError("Unable to check staged changes.")

# =============================================================================
# 07 | AUTHENTICATE AND PUSH TO GITHUB
# =============================================================================

# Use the regular GitHub HTTPS remote without storing the token in its URL.
repository_url = (
    f"https://github.com/{USERNAME}/{REPO_NAME}.git"
)

# Set the repository's remote URL.
subprocess.run(
    ["git", "remote", "set-url", "origin", repository_url],
    cwd=PROJECT_DIR,
    check=True
)

# Supply credentials to Git without displaying the token in notebook output.
askpass_script = PROJECT_DIR / ".git" / "colab_askpass.sh"

askpass_script.write_text(
    '#!/bin/sh\n'
    'case "$1" in\n'
    '  *Username*) printf "%s\\n" "$GITHUB_USERNAME" ;;\n'
    '  *Password*) printf "%s\\n" "$GITHUB_TOKEN" ;;\n'
    'esac\n'
)

askpass_script.chmod(0o700)

git_env = os.environ.copy()
git_env["GIT_ASKPASS"] = str(askpass_script)
git_env["GIT_TERMINAL_PROMPT"] = "0"
git_env["GITHUB_USERNAME"] = USERNAME
git_env["GITHUB_TOKEN"] = TOKEN

print("\nPUSHING REPORTS TO GITHUB")

try:
    subprocess.run(
        ["git", "push", "origin", BRANCH],
        cwd=PROJECT_DIR,
        env=git_env,
        check=True
    )
finally:
    # Remove temporary authentication helper and credentials.
    askpass_script.unlink(missing_ok=True)
    git_env.pop("GITHUB_TOKEN", None)
    TOKEN = None

# =============================================================================
# 08 | VERIFY SUCCESSFUL PUBLICATION
# =============================================================================

print("\n" + "=" * 55)
print("BUSINESS ANALYTICS REPORTS — GITHUB PUBLICATION")
print("=" * 55)
print("Status: Successfully pushed to GitHub")
print(f"Repository: {USERNAME}/{REPO_NAME}")
print(f"Branch: {BRANCH}")
print(f"Reports processed: {len(csv_files)}")
print(f"Output directory: {OUTPUT_DIR}")
print("=" * 55)

print(
    f"\nView reports:\n"
    f"https://github.com/{USERNAME}/{REPO_NAME}/tree/{BRANCH}/outputs"
)
````

````python
!pip install reportlab
# =============================================================================
# 07 | GENERATE EXECUTIVE BUSINESS REPORT
# =============================================================================
# Business objective:
# Transform project budget and delivery results into a concise,
# management-ready PDF with prioritized business decisions.
#
# Output:
# outputs/management_decision_brief.pdf
# =============================================================================

from pathlib import Path
import pandas as pd

from reportlab.platypus import (
    SimpleDocTemplate, Paragraph, Table, TableStyle,
    Spacer, KeepTogether
)
from reportlab.lib import colors
from reportlab.lib.styles import ParagraphStyle
from reportlab.lib.enums import TA_CENTER, TA_LEFT
from reportlab.lib.units import inch
from reportlab.lib.pagesizes import A4

# =============================================================================
# 01 | CONFIGURE FILE PATHS
# =============================================================================

PROJECT_DIR = Path("/content/python-business-analytics-project")
OUTPUT_DIR = PROJECT_DIR / "outputs"
OUTPUT_DIR.mkdir(parents=True, exist_ok=True)

REPORT_PATH = OUTPUT_DIR / "management_decision_brief.pdf"
REPORT_DATE = "31 January 2026"

# =============================================================================
# 02 | LOAD BUSINESS ANALYTICS RESULTS
# =============================================================================
# Reuse existing pipeline outputs rather than hardcoding business results.

budget = pd.read_csv(OUTPUT_DIR / "budget_by_project.csv")
departments = pd.read_csv(OUTPUT_DIR / "budget_by_department.csv")
categories = pd.read_csv(OUTPUT_DIR / "budget_by_category.csv")
delivery = pd.read_csv(OUTPUT_DIR / "project_delivery.csv")
kpis = pd.read_csv(OUTPUT_DIR / "delivery_kpis.csv").iloc[0]

# =============================================================================
# 03 | CALCULATE MANAGEMENT KPIs
# =============================================================================
# Summarize financial exposure and project delivery performance.

planned = budget["planned_budget"].sum()
actual = budget["actual_cost"].sum()
variance = actual - planned
variance_pct = variance / planned * 100 if planned else 0

over_budget = int((budget["variance"] > 0).sum())
total_projects = len(budget)

on_time = float(kpis["on_time_completed_pct"])
overdue_count = int(kpis["open_overdue_projects"])

# Identify departments with the largest unfavorable budget variances.
top_departments = (
    departments.sort_values("variance", ascending=False)
    .head(2)
)

# Identify the projects contributing the largest cost overruns.
top_cost_projects = (
    budget.sort_values("variance", ascending=False)
    .head(3)
)

# Identify the most significant net cost-category overruns.
cost_drivers = (
    categories.groupby("cost_category", as_index=False)["variance"]
    .sum()
    .sort_values("variance", ascending=False)
    .head(2)
)

# Prioritize ongoing projects with the longest overdue periods.
overdue_projects = (
    delivery.loc[delivery["overdue_open"] == True]
    .sort_values("open_days_overdue", ascending=False)
    .head(3)
)

def fmt(value):
    """Format financial values without assuming a currency."""
    return f"{value:,.0f}"

def project_summary(df, amount_col, suffix=""):
    """Convert ranked project findings into a readable summary."""
    return ", ".join(
        f"{row.project_id} ({fmt(getattr(row, amount_col))}{suffix})"
        for row in df.itertuples()
    )

department_summary = ", ".join(
    f"{row.department} (+{fmt(row.variance)})"
    for row in top_departments.itertuples()
)

category_summary = ", ".join(
    f"{row.cost_category} (+{fmt(row.variance)})"
    for row in cost_drivers.itertuples()
)

cost_project_summary = project_summary(
    top_cost_projects, "variance"
)

overdue_summary = project_summary(
    overdue_projects, "open_days_overdue", " days"
)

# =============================================================================
# 04 | DEFINE PROFESSIONAL REPORT DESIGN
# =============================================================================
# Use a restrained corporate palette and consistent typography.

NAVY = colors.HexColor("#17324D")
BLUE = colors.HexColor("#2471A3")
LIGHT_BLUE = colors.HexColor("#EAF2F8")
LIGHT_GREY = colors.HexColor("#F4F6F8")
DARK_GREY = colors.HexColor("#344054")
MID_GREY = colors.HexColor("#667085")
RED = colors.HexColor("#B54736")
WHITE = colors.white

styles = {
    "title": ParagraphStyle(
        "ReportTitle",
        fontName="Helvetica-Bold",
        fontSize=20,
        leading=25,
        textColor=NAVY,
        spaceAfter=7
    ),
    "subtitle": ParagraphStyle(
        "ReportSubtitle",
        fontName="Helvetica",
        fontSize=9,
        leading=13,
        textColor=MID_GREY,
        spaceAfter=15
    ),
    "heading": ParagraphStyle(
        "SectionHeading",
        fontName="Helvetica-Bold",
        fontSize=11,
        leading=15,
        textColor=NAVY,
        spaceBefore=14,
        spaceAfter=7
    ),
    "body": ParagraphStyle(
        "ReportBody",
        fontName="Helvetica",
        fontSize=8.5,
        leading=13,
        textColor=DARK_GREY
    ),
    "small": ParagraphStyle(
        "ReportSmall",
        fontName="Helvetica",
        fontSize=7.5,
        leading=11,
        textColor=MID_GREY
    ),
    "kpi_label": ParagraphStyle(
        "KPILabel",
        fontName="Helvetica",
        fontSize=7,
        leading=10,
        alignment=TA_CENTER,
        textColor=MID_GREY
    ),
    "kpi_value": ParagraphStyle(
        "KPIValue",
        fontName="Helvetica-Bold",
        fontSize=16,
        leading=21,
        alignment=TA_CENTER,
        textColor=NAVY
    )
}

# =============================================================================
# 05 | CREATE THE PDF DOCUMENT
# =============================================================================

doc = SimpleDocTemplate(
    str(REPORT_PATH),
    pagesize=A4,
    rightMargin=40,
    leftMargin=40,
    topMargin=40,
    bottomMargin=45
)

story = []
page_width = A4[0] - 80

# =============================================================================
# 06 | REPORT HEADER AND EXECUTIVE SUMMARY
# =============================================================================

story.append(Paragraph(
    "MANAGEMENT DECISION BRIEF",
    styles["title"]
))

story.append(Paragraph(
    f"Business Performance & Growth Opportunities | {REPORT_DATE}",
    styles["subtitle"]
))

story.append(Paragraph(
    "<b>Executive Summary</b> | "
    "The analysis highlights opportunities to strengthen budget control, "
    "recover delayed projects, and improve resource allocation. "
    "Management should prioritize the largest financial and delivery risks "
    "while evaluating commercially focused digital initiatives.",
    styles["body"]
))

story.append(Spacer(1, 14))

# =============================================================================
# 07 | EXECUTIVE KPI DASHBOARD
# =============================================================================

kpi_data = [[
    Paragraph("BUDGET VARIANCE", styles["kpi_label"]),
    Paragraph("ON-TIME DELIVERY", styles["kpi_label"]),
    Paragraph("OVERDUE PROJECTS", styles["kpi_label"])
], [
    Paragraph(f"+{variance_pct:.2f}%", styles["kpi_value"]),
    Paragraph(f"{on_time:.1f}%", styles["kpi_value"]),
    Paragraph(str(overdue_count), styles["kpi_value"])
], [
    Paragraph(f"+{fmt(variance)} above plan", styles["kpi_label"]),
    Paragraph("Completed projects", styles["kpi_label"]),
    Paragraph("Open projects", styles["kpi_label"])
]]

kpi_table = Table(
    kpi_data,
    colWidths=[page_width / 3] * 3,
    rowHeights=[21, 31, 20]
)

kpi_table.setStyle(TableStyle([
    ("BACKGROUND", (0, 0), (-1, -1), LIGHT_BLUE),
    ("BOX", (0, 0), (-1, -1), 0.5, colors.HexColor("#D7E5F0")),
    ("VALIGN", (0, 0), (-1, -1), "MIDDLE"),
    ("LEFTPADDING", (0, 0), (-1, -1), 9),
    ("RIGHTPADDING", (0, 0), (-1, -1), 9)
]))

story.append(kpi_table)

# =============================================================================
# 08 | KEY BUSINESS FINDINGS
# =============================================================================

story.append(Paragraph("Key Business Findings", styles["heading"]))

findings = [
    (
        "Financial performance",
        f"Actual spending exceeded the planned budget by {fmt(variance)} "
        f"({variance_pct:.2f}%). {over_budget} of {total_projects} projects "
        "recorded unfavorable budget variances."
    ),
    (
        "Management cost priorities",
        f"Highest departmental overruns: {department_summary}. "
        f"Main net cost drivers: {category_summary}."
    ),
    (
        "Project delivery",
        f"Only {on_time:.1f}% of completed projects were delivered on time. "
        f"{overdue_count} open projects are overdue."
    )
]

for label, description in findings:
    story.append(Paragraph(
        f"<b>{label}:</b> {description}",
        styles["body"]
    ))
    story.append(Spacer(1, 6))

# =============================================================================
# 09 | PRIORITIZED MANAGEMENT DECISIONS
# =============================================================================

story.append(Paragraph(
    "Recommended Management Decisions",
    styles["heading"]
))

recommendations = [
    (
        "01 | Recover delayed projects",
        f"Review {overdue_summary}. "
        "Confirm delivery blockers, assign accountable owners, "
        "and approve revised completion plans within 14 days.",
        "Project Management / 14 days"
    ),
    (
        "02 | Strengthen cost control",
        f"Prioritize budget reviews for {cost_project_summary}. "
        "Validate staffing costs, scope changes, and purchase approvals. "
        "Introduce monthly variance reviews.",
        "Finance & Department Leads / 30 days"
    ),
    (
        "03 | Improve resource efficiency",
        f"Investigate {category_summary}. "
        "Review resource utilization, software licenses, and vendor terms. "
        "Update forecasts based on verified cost drivers.",
        "Finance, IT & Procurement / 60 days"
    ),
    (
        "04 | Evaluate revenue opportunities",
        "Pilot CRM and service-portal improvements aimed at better "
        "customer conversion, retention, or service capacity. "
        "Measure incremental revenue and contribution margin before scaling.",
        "Sales, Marketing & Finance / 90 days"
    )
]

table_rows = [[
    Paragraph("<b>PRIORITY & ACTION</b>", styles["small"]),
    Paragraph("<b>OWNER / TIMING</b>", styles["small"])
]]

for title, action, owner in recommendations:
    table_rows.append([
        Paragraph(
            f"<b>{title}</b><br/>{action}",
            styles["body"]
        ),
        Paragraph(owner, styles["small"])
    ])

decision_table = Table(
    table_rows,
    colWidths=[page_width * 0.75, page_width * 0.25],
    repeatRows=1,
    hAlign="LEFT"
)

decision_table.setStyle(TableStyle([
    ("BACKGROUND", (0, 0), (-1, 0), LIGHT_BLUE),
    ("ROWBACKGROUNDS", (0, 1), (-1, -1), [WHITE, LIGHT_GREY]),
    ("VALIGN", (0, 0), (-1, -1), "TOP"),
    ("LEFTPADDING", (0, 0), (-1, -1), 9),
    ("RIGHTPADDING", (0, 0), (-1, -1), 9),
    ("TOPPADDING", (0, 0), (-1, -1), 8),
    ("BOTTOMPADDING", (0, 0), (-1, -1), 8),
    ("LINEBELOW", (0, -1), (-1, -1), 0.5,
     colors.HexColor("#D0D5DD"))
]))

story.append(decision_table)

# =============================================================================
# 10 | BUSINESS IMPROVEMENT METRICS
# =============================================================================

story.append(Paragraph(
    "Management Performance Dashboard",
    styles["heading"]
))

story.append(Paragraph(
    "<b>Track monthly:</b> Budget variance (%) | "
    "On-time delivery (%) | Overdue project count | "
    "Labor and software cost variance | "
    "Customer conversion (%) | Incremental contribution margin.",
    styles["body"]
))

story.append(Spacer(1, 10))

# =============================================================================
# 11 | ANALYSIS LIMITATIONS
# =============================================================================
# Distinguish measured results from proposed revenue improvements.

story.append(Paragraph(
    "<b>Evidence & limitations:</b> Findings are based on the project's "
    "budget and delivery data. The dataset does not contain sales, "
    "revenue, or profit records; proposed growth initiatives therefore "
    "require separate commercial validation. Budget and schedule "
    "variances do not establish their underlying causes. "
    "Financial units follow the source data; no currency is assumed.",
    styles["small"]
))

# =============================================================================
# 12 | ADD FOOTER AND EXPORT REPORT
# =============================================================================

def add_footer(canvas, document):
    """Add consistent branding and page numbering."""
    canvas.saveState()

    width, height = A4

    canvas.setStrokeColor(
        colors.HexColor("#D0D5DD")
    )
    canvas.line(40, 33, width - 40, 33)

    canvas.setFont("Helvetica", 7)
    canvas.setFillColor(MID_GREY)

    canvas.drawString(
        40, 22,
        "BUSINESS ANALYTICS | MANAGEMENT REPORT"
    )

    canvas.drawRightString(
        width - 40, 22,
        f"Page {document.page}"
    )

    canvas.restoreState()

doc.build(
    story,
    onFirstPage=add_footer,
    onLaterPages=add_footer
)

# =============================================================================
# 13 | CONFIRM REPORT GENERATION
# =============================================================================

print("=" * 60)
print("MANAGEMENT BUSINESS REPORT CREATED SUCCESSFULLY")
print("=" * 60)
print(f"Report: {REPORT_PATH}")
print(f"Budget variance: {variance_pct:+.2f}%")
print(f"On-time delivery: {on_time:.1f}%")
print(f"Overdue projects: {overdue_count}")
print("=" * 60)
````


## GitHub Authentication Troubleshooting

If the GitHub push fails, review the following common issues.

### Possible Causes

1. **Invalid Personal Access Token (PAT):** The token may be incorrect, expired, or revoked.
2. **Insufficient Permissions:** Classic tokens require the `repo` scope; fine-grained tokens require **Contents: Read and write** access.
3. **Missing Repository Access:** Your GitHub account must have write permission for the target repository.

#### How to Resolve

1. Navigate to **GitHub → Settings → Developer settings → Personal access tokens**.
2. Generate a new token with the required repository permissions.
3. Copy the token and rerun the Google Colab authentication cell.
4. Enter the new token when prompted and retry the GitHub push.

> **Security Note:** GitHub account passwords cannot be used for HTTPS Git operations. Never store Personal Access Tokens in notebooks, source code, or public repositories.
